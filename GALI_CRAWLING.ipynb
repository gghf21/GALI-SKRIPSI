```python
import os
import time
import shutil

twitter_auth_token = "79744989f8d8037d9cbaf2331ca3ab9997c5d7b0"

output_folder = "tweets-data"
os.makedirs(output_folder, exist_ok=True)

since_date = "2025-01-06"
until_date = "2025-11-04"
limit = 10000
language = "lang:id"

query_MBG = f'(Keracunan MBG {language}) since:{since_date} until:{until_date}'
file_MBG = "MBG6.csv"


def crawl_data(filename, search_query):
    """Menjalankan TweetHarvest dan menyimpan hasil CSV"""
    print(f"\n Mengambil data untuk: {filename} ...")
    try:
        !npx -y tweet-harvest@2.6.1 -o "{filename}" -s "{search_query}" --tab "LATEST" -l {limit} --token {twitter_auth_token}

        if os.path.exists(filename):
            shutil.move(filename, f"{output_folder}/{filename}")
            print(f"Selesai: {filename} (disimpan di folder {output_folder}/)\n")
        else:
            print(f"Gagal menemukan file output: {filename}. Mungkin token atau koneksi bermasalah.\n")
    except Exception as e:
        print(f"Terjadi error saat mengambil data {filename}: {e}\n")

    time.sleep(5)

crawl_data(file_MBG, query_MBG)

print("\n Semua data aspek berhasil diproses. Cek folder 'tweets-data' untuk hasil CSV.")
```
